```python
import sqlite3
import pandas as pd
from pathlib import Path

CAMINHO_BANCO = Path("../../dados/banco/compras.db")

```

```python
conn = sqlite3.connect(
    f"{CAMINHO_BANCO.resolve().as_uri()}?mode=ro",
    uri=True
)

df_contratacoes = pd.read_sql_query(
    "SELECT * FROM contratacoes",
    conn
)

print(f"Dimensões: {df_contratacoes.shape}")
display(df_contratacoes.head())
```

```python
consulta = """
    SELECT
        c.*,
        cc.cliente,
        i.material_ou_servico,
        i.tipo_beneficio_nome
    FROM contratacoes AS c
    LEFT JOIN contratacao_cliente AS cc
        ON c.id_interno = cc.id_interno
    LEFT JOIN itens AS i
        ON c.id_interno = i.id_interno
    ORDER BY c.id_interno    
"""

with sqlite3.connect(f"{CAMINHO_BANCO.resolve().as_uri()}?mode=ro",
    uri=True
) as conn:
    df_contratacoes = pd.read_sql_query(consulta, conn)
print(f"Dimensões:", df_contratacoes.shape)
display(df_contratacoes.head())
```
